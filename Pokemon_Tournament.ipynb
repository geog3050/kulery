```python
def import_data(filename):
    data = []
    with open(filename, 'r') as f:
        for line in f:
            parts = line.strip().split(',')
            name = parts[0]
            ptype = parts[1]
            hp = float(parts[2])
            dmg = float(parts[3])
            data.append([name, ptype, hp,dmg])
    return data        
```
